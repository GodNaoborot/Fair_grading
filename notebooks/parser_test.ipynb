```python
import re

import pandas as pd
import pickle

from pandas import read_pickle

df = pd.read_excel('Grades.xls',sheet_name="23-24 1к 1п")
to_drop = []
for key in df.keys():
    i = 1
    while not isinstance(df[key][i],str):
        i+=1
        if i == len(df[key]):
            break
    if i==len(df[key]):
        to_drop.append(key)
    elif df[key][i][0] in '+-':
        to_drop.append(key)
#print(to_drop)
df.drop(to_drop,axis=1,inplace=True)
df
```

```python
cell = df['Аналитическая геометрия'][1]
dates = [int(j.split(".")[1]) for j in re.findall(r'\((.*?)\)',cell)]
grade_cell = [j[0] for j in re.findall(r'.\(',cell)]
print(dates)
print(grade_cell)
j = 1
while grade_cell[j-1] not in '345'and j<len(grade_cell):
    j += 1
if j>=len(grade_cell): date_2 = []
else: date_2 = [dates[j]]
dates_valid = [dates[0]]+date_2
print(dates_valid)
```

```python
df_semester1 = {}
df_semester2 = {}
for key in df.keys():
    i = 1
    while not isinstance(df[key][i],str):
        i+=1
    dates = [int(j.split(".")[1]) for j in re.findall(r'\((.*?)\)', df[key][i])]
    grade_cell = [j[0] for j in re.findall(r'.\(',df[key][i])]
    j = 1
    while grade_cell[j-1] not in '345'and j<len(grade_cell):
        j += 1
    if j>=len(grade_cell): date_2 = []
    else: date_2 = [dates[j]]
    dates_valid = [dates[0]]+date_2
    #print(dates)
    if any(item in dates_valid for item in (11,12,1,2,3,4)):
        df_semester1[key] = []
        for cell in df[key]:
            #print(cell)
            if isinstance(cell,float):
                df_semester1[key].append(cell)
            else:
                if len(cell)<2:
                    df_semester1[key].append(float("nan"))
                    continue
                grade_cell = [j[0] for j in re.findall(r'.\(',cell)]
                #print(grade_cell)
                j = 0
                while j<len(grade_cell) and grade_cell[j] not in '345':
                    j += 1
                j = min(j,len(grade_cell)-1)
                if grade_cell[j].isdigit():
                    df_semester1[key].append(int(grade_cell[j]))
                else:
                    df_semester1[key].append(float("nan"))
    if any(item in dates for item in (5,6,7,8,9,10)):
        df_semester2[key] = []
        for cell in df[key]:
            if isinstance(cell,float):
                df_semester2[key].append(cell)
            else:
                if len(cell)<2:
                    df_semester2[key].append(float("nan"))
                    continue
                grade_cell = [j[0] for j in re.findall(r'.\(',cell)]
                #print(len(cell),cell)
                if grade_cell[-1].isdigit():
                    df_semester2[key].append(int(grade_cell[-1]))
                else:
                    df_semester2[key].append(float("nan"))

df_semester1 = pd.DataFrame(df_semester1)
df_semester2 = pd.DataFrame(df_semester2)

data = {"semester1":df_semester1,"semester2":df_semester2}

with open('23-24_1c1p.pkl', 'wb') as file:
    pickle.dump(data, file)

data = {}

with open('23-24_1c1p.pkl', 'rb') as file:
    data = pickle.load(file)
#print(df_semester1)
#print(df_semester2)
```
